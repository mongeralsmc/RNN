```python
import random
import math

class SimpleRNN:
    def __init__(self, vocab_size, hidden_size):
        # Initialize parameters
        self.vocab_size = vocab_size
        self.hidden_size = hidden_size
        
        # Embedding weights (vocab_size x hidden_size)
        self.W_embed = [[random.random() - 0.5 for _ in range(hidden_size)] 
                       for _ in range(vocab_size)]
        
        # Recurrent weights (hidden_size x hidden_size)
        self.W_recur = [[random.random() - 0.5 for _ in range(hidden_size)] 
                       for _ in range(hidden_size)]
        
        # Output weights (hidden_size x vocab_size)
        self.W_out = [[random.random() - 0.5 for _ in range(vocab_size)] 
                     for _ in range(hidden_size)]
        
        # Hidden state
        self.h = [0.0 for _ in range(hidden_size)]
    
    def forward(self, inputs):
        """Process a sequence of inputs and return final output"""
        self.h = [0.0 for _ in range(self.hidden_size)]  # Reset hidden state
        
        for x in inputs:
            # Embedding layer
            embedded = [0.0 for _ in range(self.hidden_size)]
            for i in range(self.hidden_size):
                embedded[i] = self.W_embed[x][i]
            
            # Recurrent layer
            new_h = [0.0 for _ in range(self.hidden_size)]
            for i in range(self.hidden_size):
                sum_recur = 0.0
                for j in range(self.hidden_size):
                    sum_recur += self.h[j] * self.W_recur[j][i]
                new_h[i] = math.tanh(embedded[i] + sum_recur)
            self.h = new_h
        
        # Output layer
        output = [0.0 for _ in range(self.vocab_size)]
        for i in range(self.vocab_size):
            for j in range(self.hidden_size):
                output[i] += self.h[j] * self.W_out[j][i]
        
        # Softmax
        exp_output = [math.exp(o) for o in output]
        sum_exp = sum(exp_output)
        softmax_output = [e/sum_exp for e in exp_output]
        
        return softmax_output
    
    def train(self, sequences, targets, learning_rate=0.1, epochs=100):
        """Train the RNN on sequences and targets"""
        for epoch in range(epochs):
            total_loss = 0.0
            for seq, target in zip(sequences, targets):
                # Forward pass
                output = self.forward(seq)
                
                # Calculate loss (cross-entropy)
                loss = -math.log(output[target])
                total_loss += loss
                
                # Backward pass (simplified)
                # Calculate output gradient
                output_grad = [0.0 for _ in range(self.vocab_size)]
                for i in range(self.vocab_size):
                    output_grad[i] = output[i] - (1 if i == target else 0)
                
                # Update output weights
                for i in range(self.hidden_size):
                    for j in range(self.vocab_size):
                        self.W_out[i][j] -= learning_rate * output_grad[j] * self.h[i]
                
                # Simplified backprop through time (not full BPTT)
                # Update embedding and recurrent weights based on final hidden state
                for i in range(self.vocab_size):
                    for j in range(self.hidden_size):
                        self.W_embed[i][j] -= learning_rate * output_grad[target] * (1 - self.h[j]**2) * self.W_out[j][target]
                
                for i in range(self.hidden_size):
                    for j in range(self.hidden_size):
                        self.W_recur[i][j] -= learning_rate * output_grad[target] * (1 - self.h[j]**2) * self.W_out[j][target]
            
            if epoch % 10 == 0:
                print(f"Epoch {epoch}, Loss: {total_loss/len(sequences)}")

def prepare_data(text):
    """Convert text to numerical sequences"""
    words = text.split()
    vocab = list(set(words))
    word_to_idx = {word: i for i, word in enumerate(vocab)}
    
    # Create sequences of 3 words as input and 4th word as target
    sequences = []
    targets = []
    for i in range(len(words) - 3):
        seq = [word_to_idx[words[i+j]] for j in range(3)]
        target = word_to_idx[words[i+3]]
        sequences.append(seq)
        targets.append(target)
    
    return sequences, targets, word_to_idx, vocab

# Example usage
text = "I love neural networks and machine learning is fascinating"
sequences, targets, word_to_idx, vocab = prepare_data(text)

# Create and train RNN
rnn = SimpleRNN(vocab_size=len(vocab), hidden_size=10)
rnn.train(sequences, targets, learning_rate=0.1, epochs=100)

# Test prediction
test_seq = [word_to_idx["love"], word_to_idx["neural"], word_to_idx["networks"]]
output = rnn.forward(test_seq)
predicted_idx = output.index(max(output))
print(f"Input: {' '.join([vocab[idx] for idx in test_seq])}")
print(f"Predicted next word: {vocab[predicted_idx]}")
```
